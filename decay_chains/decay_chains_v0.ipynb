```python
pdg_id_dicc = {13: r"$\mu^-$", 111: r"$\pi^0$", 321: "$K^+$", 2212: r"$p^+$"}

def text2list(text):
    if text == "":
        print(f"Este tambien venia vacio: {text}")
        return []
    
    text = text[1:-2]
    l = text.split(",")

    if l == ['']:
        print(f"Este venia vacio: {text}")
        return []

    try:
        l = [int(x) for x in l]
    except:
        print(f"Exception ocurred with: {l}")

    return l
```
